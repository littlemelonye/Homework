```python
import numpy as np
from scipy.stats import norm
from scipy.optimize import root

VE = 3
sigmaE = 0.30
X = 10
T = 2
r = 0.05

def equations(z):

    VA, sigmaA = z

    d1 = (
        np.log(VA / X)
        + (r + 0.5 * sigmaA**2) * T
    ) / (sigmaA * np.sqrt(T))

    d2 = d1 - sigmaA * np.sqrt(T)

    eq1 = (
        VA * norm.cdf(d1)
        - X * np.exp(-r * T) * norm.cdf(d2)
        - VE
    )

    eq2 = (
        VA / VE
        * norm.cdf(d1)
        * sigmaA
        - sigmaE
    )

    return [eq1, eq2]


sol = root(
    equations,
    x0=[12, 0.08]
)

VA_hat, sigmaA_hat = sol.x

print("资产价值 =", VA_hat)
print("资产波动率 =", sigmaA_hat)
```
